```python
import json
from pathlib import Path
from urllib.request import urlopen

from aads_public.evidence import load_acceptance_report

url = "https://raw.githubusercontent.com/EfeErim/bitirmeprojesi/master/evidence/controlled_demo_summary.json"
payload = json.loads(urlopen(url, timeout=30).read())
path = Path("controlled_demo_summary.json")
path.write_text(json.dumps(payload), encoding="utf-8")
report = load_acceptance_report(path)
assert report.controlled_demo_passed
report
```

```python
print(f"✅ Controlled acceptance: {report.passed}/{report.total}")
print(f"✅ Correct disease answers: {report.answered}/{report.answered}")
print(f"🛡️ Safe review/abstain: {report.reviewed}/{report.reviewed}")
print(f"🛡️ Negative false accepts: {report.negative_false_accepts}")
print("Scope: controlled demo · NOT production-ready")
```
