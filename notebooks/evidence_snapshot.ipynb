```python
from pathlib import Path
from urllib.request import Request, urlopen

from aads_public.evidence import load_acceptance_report

tag = "aads-public-demo-v1.1.1"
base = f"https://raw.githubusercontent.com/EfeErim/bitirmeprojesi/{tag}/evidence"
headers = {"User-Agent": "aads-public-notebook/1.1"}
for name in ("controlled_demo_summary.json", "controlled_demo_rows.json"):
    content = urlopen(Request(f"{base}/{name}", headers=headers), timeout=30).read()
    Path(name).write_bytes(content)

report = load_acceptance_report("controlled_demo_summary.json", "controlled_demo_rows.json")
assert report.controlled_demo_passed
print("Manifest identity: PASS")
print(f"Sanitized decisions: {report.passed}/{report.total} PASS")
print(f"Disease answers: {report.answered}")
print(f"Safe review/abstain: {report.reviewed}")
print(f"Production ready: {report.production_ready}")
```

The validation above proves the integrity and internal consistency of one recorded controlled-demo run. It does not prove field performance; all eight independent production-readiness gates remain failed.
